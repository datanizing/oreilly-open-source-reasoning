# OpenRouter

OpenRouter is a commercial service which hosts a multitude of models.
In order to use it, you have to register, fill your account with some
money. Afterwards, you can directly chat with the models or create
an API key to access it programmatically. OpenRouter directs you t
the cheapest provider of the respective model.

Why is OpenRouter useful?
* Access to many models, even those which are too large for your hardware.
* Access to the lastest models, sometimes also before integration to `llama.cpp` or `transformers` is available.
* Identical API for all models.
* API very similar to Open AI API.

```python
from openrouter import OpenRouter
import os
```

You can create the client by passing an API key. Of course, you can
also integrate the key directly into this notebook (obviously not
useful for the live course 😉).

```python
client = OpenRouter(api_key=os.getenv("OPENROUTER_API_KEY"))
```

This function lists all models which are available (very long list):

```python
models = client.models.list()
models
```

```python
import pandas as pd
mdf = pd.DataFrame([[m.id, m.name, m.context_length, m.pricing.prompt, m.pricing.completion] 
                     for m in models.data], columns=["id", "name", "Context Length", "Prompt Parsing", "Completion"])
mdf
```

```python
mdf[mdf["id"].str.contains("mimo")]
```

API is quite similar to Open AI:

```python
response = client.chat.send(
    model="xiaomi/mimo-v2.5-pro",
    messages=[{"role": "user", "content": "How many 'r's are in 'strawberry'?"}]
)
```

Result structure is identical:

```python
response.choices
```

Reasoning and content are nicely separated:

```python
from IPython.display import display, Markdown
display(Markdown(response.choices[0].message.reasoning))
```

```python
display(Markdown(response.choices[0].message.content))
```

## Try a different model

```python
def answer_with_reasoning(model, question):
    response = client.chat.send(
        model=model,
        messages=[{"role": "user", "content": question}]
    )
    display(Markdown("### Reasoning"))
    display(Markdown(response.choices[0].message.reasoning))
    display(Markdown("### Answer"))
    display(Markdown(response.choices[0].message.content))
```

```python
mdf[mdf["id"].str.contains("glm")]
```

```python
answer_with_reasoning("z-ai/glm-5.1", car_wash)
```
