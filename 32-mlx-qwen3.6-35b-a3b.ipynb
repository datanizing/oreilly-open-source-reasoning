# Qwen3.6-35B-A3B model with `mlx`

Qwen3.6-35B-A3B is a much larger model, but due to its MoE architecture, only a fraction of the 
parameters are used during inference.

If you have the necessary RAM (which is not too unlikely in Macs, as this is combined
CPU/GPU RAM), this model is faster than even Qwen3-4B while being of course much more
powerful.

```python
from mlx_lm import load, generate
```

```python
model, tokenizer = load("mlx-community/Qwen3.6-35B-A3B-4bit-DWQ")
```

```python
prompt = "How many 'r's are in 'strawberry'?"
```

```python
if tokenizer.chat_template is not None:
    messages = [{"role": "user", "content": prompt}]
    prompt = tokenizer.apply_chat_template(
        messages, add_generation_prompt=True
    )
```

```python
response = generate(model, tokenizer, prompt=prompt, verbose=True, max_tokens=2500)
```

```python
print(response)
```

```python
from IPython.display import display, Markdown
display(Markdown(response))
```
